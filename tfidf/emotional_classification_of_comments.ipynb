```python
import pandas as pd
df = pd.read_csv('Emotion_classify_Data.csv')
df.head()
```

```python
df['Emotion'].value_counts()
```

```python
df['output'] = df['Emotion'].map({'joy': 0, 'fear': 1, 'anger': 2})
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.Comment, 
    df.output, 
    test_size=0.2, # 20% samples will go to test dataset
    random_state=2022,
    stratify=df.output
)
```

```python
print("Shape of X_train: ", X_train.shape)
print("Shape of X_test: ", X_test.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
clf = Pipeline([('vectorizer', CountVectorizer(ngram_range = (3, 3))), ('classifier', RandomForestClassifier())])
```

```python
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
clf = Pipeline([
    ('vectorizer_bi_grams', CountVectorizer(ngram_range = (1, 2))),                       #using the ngram_range parameter 
    ('random_forest', (RandomForestClassifier()))         
])
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
clf = Pipeline([
    ('vectorizer_bi_grams', CountVectorizer()),                       #using the ngram_range parameter 
    ('random_forest', (RandomForestClassifier()))         
])
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print(classification_report(y_test, y_pred))
```
